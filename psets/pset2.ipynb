## Problem 1: Fourier Stability of Multi-Step and Multi-Stage Methods (5+10+10)

The power of Fourier analysis is that it reduces stability analysis of a PDE discretization to: (1) finding $\lambda_m \Delta t$ as a function of $\beta_m = k_m \Delta x$ for the spatial scheme, and (2) checking that $\lambda_m \Delta t$ lies inside the stability region of the time integrator for all $\beta_m \in [-\pi, \pi]$.

In this problem we fix the spatial discretization and ask: *how does the choice of time integrator change the allowable CFL?*

Notation. Throughout this problem, $\nu = u\Delta t/\Delta x$ denotes the CFL number.

**1(a) First-order upwinding (5 pts)** 

Consider the first-order upwind spatial discretization of the 1D convection equation ($\mu = 0$, $u > 0$):

$$\frac{dU_j}{dt} = -\frac{u}{\Delta x}(U_j - U_{j-1}).$$

The Fourier eigenvalues are $\lambda_m = -\frac{u}{\Delta x}(1 - e^{-i\beta_m})$.

For each of the following time integrators, derive the amplification factor $g(\beta_m) \equiv \hat{U}_m^{n+1}/\hat{U}_m^n$:

1. Forward Euler: $U^{n+1} = U^n + \Delta t \, f(U^n)$
2. Leapfrog (second-order centered in time): $U^{n+1} = U^{n-1} + 2\Delta t \, f(U^n)$
3. RK2 (Heun's method): $U^* = U^n + \Delta t\,f(U^n)$, $\;U^{n+1} = U^n + \frac{\Delta t}{2}[f(U^n) + f(U^*)]$

For leapfrog, note that $g$ satisfies a *quadratic* equation. Find both roots and discuss which is the "physical" mode and which is the "computational" (parasitic) mode.

## Problem 2: Dispersion, Phase Speed, and Group Velocity (5+10+10)

For the semi-discrete equation $d\mathbf{U}/dt = A\mathbf{U}$ with periodic BCs, Fourier analysis gives eigenvalues $\lambda_m$ parameterized by the reduced wavenumber $\beta_m = k_m\Delta x \in [-\pi, \pi]$. The *numerical phase speed* and *group velocity* of mode $m$ are defined as follows:
$$c_{\text{phase}}(\beta) = -\frac{\operatorname{Im}(\lambda)}{k} = -\Delta x\,\frac{\operatorname{Im}(\lambda)}{\beta}, \qquad c_{\text{group}}(\beta) = -\frac{d\,\operatorname{Im}(\lambda)}{dk} = -\Delta x\,\frac{d\,\operatorname{Im}(\lambda)}{d\beta}$$
where for a specific mode $m$, one evaluates these expressions at $\lambda = \lambda_m$, $\beta = \beta_m = k_m \Delta x$, and $k = k_m$.

For the exact convection equation, $\lambda_m = -iuk_m$ yields $c_{\text{phase}} = c_{\text{group}} = u$ for *all* modes $m$, i.e., no dispersion. Any deviation from $u$ is a purely numerical artifact.
